```python
stores = pd.read_csv('data/store_locations.csv')

# clean the IDs (' FL-078', 'FL078', 'FL_287' -> 'FL-078') and drop the duplicate rows
stores['store_id'] = (stores['store_id'].str.strip().str.upper()
                      .str.replace(r'^FL[-_]?', 'FL-', regex=True))
stores = stores.drop_duplicates('store_id')

frames = []
for _, s in stores.iterrows():
    if pd.isna(s['latitude']):
        print('no coordinates for', s['store_id'], s['city'])
        continue
    w = daily_weather('2024-09-01', '2024-09-30', lat=s['latitude'], lon=s['longitude'])
    w['store_id'] = s['store_id']
    frames.append(w)
    time.sleep(1)   # be polite to the API

weather_all = pd.concat(frames, ignore_index=True)
print(weather_all.shape)   # expect 30 days x number of stores
```
